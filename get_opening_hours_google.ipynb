Code to get opening hours from google: requirres apify api token, which is limited to requests per month. 


```python
import geopandas as gpd
import pandas as pd
from apify_client import ApifyClient
```

```python
# Initialize the ApifyClient with your Apify API token
# Replace '<YOUR_API_TOKEN>' with your token.
client = ApifyClient('<YOUR_API_TOKEN>')

# Prepare the Actor input
run_input = {
    "searchStringsArray": ["cafe", "coffee shop", "restaurant"],
    "locationQuery": "Leeds, UK",
    "language": "en",
    "countryCode": "",
}

# Run the Actor and wait for it to finish
run = client.actor("compass/google-maps-extractor").call(run_input=run_input)

# Fetch and print Actor results from the run's dataset (if there are any)
print("💾 Check your data here: https://console.apify.com/storage/datasets/" + run["defaultDatasetId"])
for item in client.dataset(run["defaultDatasetId"]).iterate_items():
    print(item)
```

```python
# get cafe data in a df
cafe_data = []
for item in client.dataset(run["defaultDatasetId"]).iterate_items():
    cafe_data.append(item)

cafe_data =pd.DataFrame(cafe_data)

# Split the column into two columns: lat and long
cafe_data[['lat', 'long']] = cafe_data['location'].apply(lambda row: pd.Series([row['lat'], row['lng']]))

# convert to gpd dataframe
cafe_data =gpd.GeoDataFrame(cafe_data,geometry=gpd.points_from_xy(cafe_data.long, cafe_data.lat), crs="EPSG:4326"
)
```
